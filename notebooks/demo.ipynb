# Hologram Genuineness Detector — Demo

Runs the full classical CV pipeline (ROI → preprocessing → HSV classification → filtering → sequence compression → decision) on a sample video and inspects every intermediate artifact.

Run this from the project root with the `hologram-detector` environment active.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from app.detector import HologramDetector

detector = HologramDetector()
```

## 1. Run detection

Point `VIDEO_PATH` at a hologram tilt video, e.g. one placed under `dataset/genuine/` or `dataset/test/`.

```python
VIDEO_PATH = PROJECT_ROOT / "dataset" / "test" / "sample.mp4"  # replace with a real video

result = detector.process_video(str(VIDEO_PATH), roi_mode="auto")
print(f"Label:      {result.label}")
print(f"Sequence:   {result.sequence}")
print(f"Confidence: {result.confidence:.2%}")
print(f"Reasoning:  {result.reasoning}")
```
